```python
# SQL 전송 -> 전송결과 받아 원하는 로직
cursor = conn.cursor() # SQL 전송&결과 받는 객체
sql = "SELECT * FROM EMP ORDER BY EMPNO"
cursor.execute(sql) #SQL전송 + 결과 받기
emps = cursor.fetchall()  # 전송결과 tuple list로 모두 받기
keys=[desc[0].lower() for desc in cursor.description]
print(keys)
cursor.close()
```
