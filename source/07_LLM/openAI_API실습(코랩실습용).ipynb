```python
# 보안 비밀키를 추가 -> 액세스 가능하도록
from openai import OpenAI
from google.colab import userdata
userdata.get('OPENAI_API_KEY')[:2]
client = OpenAI(api_key=userdata.get('OPENAI_API_KEY'))
```

```python
response = client.audio.speech.create(
    model='tts-1',
    voice='nova',
    input='Hello, OpenAI!'
)
audio_data=response.content
```

```python
print(f'생성된 오디오 크기 : {len(audio_data):,}바이트')
with open('/content/drive/MyDrive/Colab Notebooks/ch5_nova.mp3','wb') as audio_file:
  audio_file.write(audio_data)
```
