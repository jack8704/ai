## 2.2 OpenAI SDK를 사용

```python
from openai import OpenAI
import os
client = OpenAI(
    api_key=os.getenv('UPSTAGE_API_KEY'),
    base_url="https://api.upstage.ai/v1"
)

response = client.chat.completions.create(
    model = "solar-pro2",
    messages=[
        {'role':'system', 'content':'너는 친절하게 대답해 주는 비서야'},
        {'role':'user','content':'2020년 월드 시리즈 누가 우승했어?'}
    ],
    temperature=0.5, # 0~0.3: 일관되고 예측가능한 답변  / 0.8~1 : 창의적이고 다양한 답변
    stream=False
)
```

```python
print(response.choices[0].message.content)
```

```python
response.usage # 사용 토큰수
```

## 2.2 LangChain 사용
- 발급받은 API KEY를 'UPSTAGE_API_KEY'라고 저장하면 별도의 설정없음

```python
from langchain_upstage import ChatUpstage
from langchain_core.messages import SystemMessage, HumanMessage
llm=ChatUpstage()
messages=[
    SystemMessage(content='너는 쉽고 친절하게 대답해주는 비서야'),
    HumanMessage(content='2020년 월드시리즈 누가 우승했어?')
]
ai_message=llm.invoke(messages)
print(ai_message.content)
```
