## 2.2 OpenAI SDK를 사용

```python
from openai import OpenAI

client = OpenAI(
    api_key="text", #필수 인자라 생략할 수 없어서 아무거나 입력
    base_url="http://localhost:11434/v1"
)

response = client.chat.completions.create(
    model = "exaone3.5:2.4b",
    messages=[
        {'role':'system', 'content':'너는 친절하게 대답해 주는 비서야'},
        {'role':'user','content':'2020년 월드 시리즈 누가 우승했어?'}
    ],
    temperature=0.5, # 0~0.3: 일관되고 예측가능한 답변  / 0.8~1 : 창의적이고 다양한 답변
    stream=False
)
```

```python
print(response.choices[0].message.content)
```

## 2.2 LangChain 사용

```python
%%time
from langchain_ollama import ChatOllama
from langchain_core.messages import SystemMessage, HumanMessage
llm=ChatOllama(
        model = 'exaone3.5:2.4b')
messages=[
    SystemMessage(content='너는 쉽고 친절하게 대답해주는 비서야'),
    HumanMessage(content='2020년 월드시리즈 누가 우승했어?')
]
ai_message=llm.invoke(messages)
print(ai_message.content)
```
